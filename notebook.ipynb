## Data Loading & Cleaning

```python
import pandas as pd

retail_df = pd.read_csv("Online_Retail.csv")
retail_df = retail_df.dropna(subset=["CustomerID"])
retail_df = retail_df[retail_df['Quantity'] >= 0]
retail_df['TotalSum'] = retail_df['Quantity'] * retail_df['UnitPrice']
retail_df['InvoiceDate'] = pd.to_datetime(retail_df['InvoiceDate'], format="%d.%m.%Y %H:%M")
print(retail_df.info())
```

## RFM Calculation

```python
import datetime as  dt

snapshot_date = max(retail_df['InvoiceDate'])+dt.timedelta(days=1)
print(snapshot_date)

data_Customer = retail_df.groupby('CustomerID').agg(Date_lost_bay=('InvoiceDate', 'max'))
data_Customer['Recency'] = (snapshot_date - data_Customer['Date_lost_bay']).dt.days
                                                    
print(data_Customer)
```
